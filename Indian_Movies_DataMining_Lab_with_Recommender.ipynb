### Imports, loading & cleaning
The raw file is messy: missing values are written as `-`, `Year` contains things like `II 2017` / `2018 Video`, `Timing(min)` is text (`134 min`), `Votes` has commas, and `Genre` is a comma-separated string. Everything is cleaned and turned into numeric features here.

```python
import io, gzip, base64, os, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gradio as gr

from scipy.cluster.hierarchy import linkage, dendrogram, cophenet
from scipy.spatial.distance import pdist
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (r2_score, mean_absolute_error, mean_squared_error, accuracy_score,
                             confusion_matrix, classification_report, silhouette_score,
                             davies_bouldin_score, calinski_harabasz_score)
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.naive_bayes import GaussianNB
from sklearn.feature_selection import mutual_info_classif
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.decomposition import PCA
from mlxtend.frequent_patterns import apriori, association_rules

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", context="notebook")
RS = 42  # random state used everywhere

# ---------- load ----------
if os.path.exists("indian_movies.csv"):
    raw = pd.read_csv("indian_movies.csv"); SRC = "uploaded indian_movies.csv"
else:
    raw = pd.read_csv(io.BytesIO(gzip.decompress(base64.b64decode("".join(DATA_B64.split())))))
    SRC = "embedded copy of indian_movies.csv"

# ---------- clean ----------
d = raw.drop_duplicates().replace("-", np.nan)
d["Year"]    = pd.to_numeric(d["Year"].astype(str).str.extract(r"(\d{4})")[0], errors="coerce")
d["Runtime"] = pd.to_numeric(d["Timing(min)"].astype(str).str.extract(r"(\d+)")[0], errors="coerce")
d["Rating"]  = pd.to_numeric(d["Rating(10)"], errors="coerce")
d["Votes"]   = pd.to_numeric(d["Votes"].astype(str).str.replace(",", ""), errors="coerce")
d["Genre"]   = d["Genre"].str.strip()
d["Language"] = d["Language"].str.strip().str.lower()
d = d[["ID", "Movie Name", "Year", "Runtime", "Rating", "Votes", "Genre", "Language"]]

MISSING_PCT = (d.isna().mean() * 100).round(2)          # for the missing-values plot
df_all = d.copy()                                        # every movie (used for Apriori / genre counts)

# rows complete enough for modelling; drop absurd runtimes (<40 or >300 min)
df = d.dropna(subset=["Year", "Runtime", "Rating", "Votes", "Genre"]).copy()
df = df[(df.Runtime >= 40) & (df.Runtime <= 300)].reset_index(drop=True)

# ---------- feature engineering ----------
df["log_Votes"]     = np.log10(df["Votes"])
df["Genre_List"]    = df["Genre"].str.split(r",\s*")
df["Num_Genres"]    = df["Genre_List"].str.len()
df["Primary_Genre"] = df["Genre_List"].str[0]
df["Decade"]        = (df["Year"] // 10 * 10).astype(int)
df["Rating_Category"] = pd.cut(df["Rating"], [0, 5.5, 7, 10.1], right=False,
                               labels=["Low (<5.5)", "Medium (5.5-7)", "High (>=7)"]).astype(str)

TOP_GENRES = list(df["Genre_List"].explode().value_counts().head(10).index)
GENRE_COLS = []
for g in TOP_GENRES:
    df["Genre_" + g] = df["Genre_List"].apply(lambda l: int(g in l)); GENRE_COLS.append("Genre_" + g)

TOP_LANGS = list(df["Language"].value_counts().head(6).index)
df["Lang_Group"] = np.where(df["Language"].isin(TOP_LANGS), df["Language"], "other")
LANG_COLS = []
for l in TOP_LANGS:
    df["Lang_" + l] = (df["Language"] == l).astype(int); LANG_COLS.append("Lang_" + l)

TOP_PG = list(df["Primary_Genre"].value_counts().head(5).index)
df["Primary_Genre_Group"] = np.where(df["Primary_Genre"].isin(TOP_PG), df["Primary_Genre"], "Other")

NUM_FEATS = ["Year", "Runtime", "log_Votes", "Num_Genres"]      # core numeric attributes
BIN_COLS  = GENRE_COLS + LANG_COLS
NUM_COLS  = ["Year", "Runtime", "Rating", "Votes", "log_Votes", "Num_Genres"]
CAT_COLS  = ["Language", "Lang_Group", "Primary_Genre", "Rating_Category", "Decade", "Num_Genres"]

print(f"Source: {SRC}")
print(f"Raw rows: {len(raw):,} | after de-dup: {len(df_all):,} | complete & cleaned (used for modelling): {len(df):,}")
print("Rating_Category balance:\n", df["Rating_Category"].value_counts(normalize=True).round(3))
```

### Shared helpers

```python
def msg_fig(text):
    fig, ax = plt.subplots(figsize=(7, 2.5)); ax.axis("off")
    ax.text(0.5, 0.5, text, ha="center", va="center", fontsize=12, wrap=True); return fig

def safe(fn):
    """Show errors in the UI instead of crashing the app. The first output must be a Markdown box."""
    def wrapper(*a, **k):
        try:
            return fn(*a, **k)
        except Exception as e:
            return (f"⚠️ **{type(e).__name__}:** {e}", *([None] * (OUT_COUNT[fn.__name__] - 1)))
    wrapper.__name__ = fn.__name__
    return wrapper

OUT_COUNT = {}   # filled per experiment below

def cm_fig(cm, labels, title, normalize=False):
    if normalize:
        cm = cm.astype(float) / cm.sum(axis=1, keepdims=True).clip(min=1)
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt=".2f" if normalize else "d", cmap="Blues", cbar=False,
                xticklabels=labels, yticklabels=labels, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual"); ax.set_title(title)
    plt.setp(ax.get_xticklabels(), rotation=30, ha="right"); fig.tight_layout(); return fig

def report_df(y_true, y_pred):
    r = pd.DataFrame(classification_report(y_true, y_pred, output_dict=True, zero_division=0)).T
    r = r.round(3).reset_index().rename(columns={"index": "class"})
    r["support"] = r["support"].astype(int); return r

def get_features(target):
    f = NUM_FEATS.copy()
    if target != "Primary_Genre_Group": f += GENRE_COLS   # avoid leaking the target into the features
    if target != "Lang_Group":          f += LANG_COLS
    return f

TARGET_CHOICES = ["Rating_Category", "Primary_Genre_Group", "Lang_Group"]
```

## Experiment 3 – Data Visualization
Twelve different plot types to understand the dataset's characteristics (distribution, relationships, trends, categories, correlations, missing data).

```python
VIZ_TYPES = ["Histogram", "Box plot", "Violin plot", "Scatter plot", "Hexbin (density)", "Line plot (trend)",
             "Bar chart", "Pie chart", "Correlation heatmap", "Pair plot", "Genre frequency", "Missing values (raw data)"]
HUE_CHOICES = ["(none)", "Lang_Group", "Rating_Category"]

def _need_num(c):
    if c not in NUM_COLS: raise ValueError(f"'{c}' is not numeric – choose one of {NUM_COLS}.")

def run_viz(kind, x, y, hue, bins, top_n):
    plt.close("all")
    try:
        h = None if hue == "(none)" else hue
        yy = None if y == "(none)" else y
        top_n = int(top_n)
        def order(col):
            if col in ("Decade", "Num_Genres"): return sorted(df[col].unique())
            return list(df[col].value_counts().head(top_n).index)

        if kind == "Histogram":
            _need_num(x); fig, ax = plt.subplots(figsize=(9, 5))
            sns.histplot(data=df, x=x, hue=h, bins=int(bins), kde=True, ax=ax); ax.set_title(f"Distribution of {x}")

        elif kind in ("Box plot", "Violin plot"):
            if yy is None: raise ValueError("Choose a numeric Y (e.g. Rating).")
            _need_num(yy); fig, ax = plt.subplots(figsize=(10, 5))
            if x in NUM_COLS and df[x].nunique() > 15: raise ValueError("X must be categorical here (Language, Primary_Genre, Rating_Category, Decade …).")
            o = order(x); dd = df[df[x].isin(o)]
            (sns.boxplot if kind == "Box plot" else sns.violinplot)(data=dd, x=x, y=yy, order=o, ax=ax)
            ax.set_title(f"{yy} by {x}"); plt.setp(ax.get_xticklabels(), rotation=35, ha="right")

        elif kind == "Scatter plot":
            if yy is None: raise ValueError("Choose a numeric Y.")
            _need_num(x); _need_num(yy); fig, ax = plt.subplots(figsize=(9, 6))
            s = df.sample(min(len(df), 5000), random_state=RS)
            sns.scatterplot(data=s, x=x, y=yy, hue=h, alpha=0.5, s=18, ax=ax); ax.set_title(f"{yy} vs {x} (sample of {len(s)})")

        elif kind == "Hexbin (density)":
            if yy is None: raise ValueError("Choose a numeric Y.")
            _need_num(x); _need_num(yy); fig, ax = plt.subplots(figsize=(9, 6))
            hb = ax.hexbin(df[x], df[yy], gridsize=35, cmap="viridis", mincnt=1)
            fig.colorbar(hb, ax=ax, label="movies"); ax.set_xlabel(x); ax.set_ylabel(yy); ax.set_title(f"Density of {yy} vs {x}")

        elif kind == "Line plot (trend)":
            if x not in ("Year", "Decade", "Runtime", "Num_Genres"): raise ValueError("Use Year or Decade as X for a trend.")
            fig, ax = plt.subplots(figsize=(10, 5))
            if yy is None:
                dd = df.groupby([x] + ([h] if h else [])).size().reset_index(name="Movies")
                sns.lineplot(data=dd, x=x, y="Movies", hue=h, marker="o", ax=ax); ax.set_title(f"Number of movies per {x}")
            else:
                _need_num(yy); sns.lineplot(data=df, x=x, y=yy, hue=h, errorbar="sd", ax=ax); ax.set_title(f"Mean {yy} per {x} (band = ±1 SD)")

        elif kind == "Bar chart":
            o = order(x); dd = df[df[x].isin(o)]; fig, ax = plt.subplots(figsize=(10, 5))
            if yy is None:
                sns.countplot(data=dd, x=x, hue=h, order=o, ax=ax); ax.set_title(f"Number of movies by {x}")
            else:
                _need_num(yy); sns.barplot(data=dd, x=x, y=yy, hue=h, order=o, errorbar=None, ax=ax); ax.set_title(f"Mean {yy} by {x}")
            plt.setp(ax.get_xticklabels(), rotation=35, ha="right")

        elif kind == "Pie chart":
            vc = df[x].value_counts(); top = vc.head(top_n)
            if len(vc) > top_n: top["Other"] = vc.iloc[top_n:].sum()
            fig, ax = plt.subplots(figsize=(7, 7)); ax.pie(top.values, labels=top.index, autopct="%1.1f%%", startangle=90)
            ax.set_title(f"Share of movies by {x}")

        elif kind == "Correlation heatmap":
            fig, ax = plt.subplots(figsize=(7, 6))
            sns.heatmap(df[NUM_COLS].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax); ax.set_title("Correlation matrix")

        elif kind == "Pair plot":
            cols = ["Year", "Runtime", "Rating", "log_Votes"]
            s = df.sample(min(len(df), 1500), random_state=RS)
            g = sns.pairplot(s, vars=cols, hue=h, corner=True, plot_kws=dict(alpha=0.4, s=12), height=2.2)
            fig = g.figure

        elif kind == "Genre frequency":
            gc = df_all["Genre"].dropna().str.split(r",\s*").explode().value_counts().head(top_n)
            fig, ax = plt.subplots(figsize=(9, 6)); sns.barplot(x=gc.values, y=gc.index, ax=ax, color="#4c72b0")
            ax.set_xlabel("Number of movies"); ax.set_title(f"Top {top_n} genres (all movies with a genre)")

        elif kind == "Missing values (raw data)":
            fig, ax = plt.subplots(figsize=(8, 5)); m = MISSING_PCT.sort_values()
            ax.barh(m.index, m.values, color="#c44e52")
            for i, v in enumerate(m.values): ax.text(v + 0.5, i, f"{v}%", va="center")
            ax.set_xlabel("% missing"); ax.set_title("Missing values per column (original file, '-' treated as missing)")
            ax.set_xlim(0, max(100, m.max() + 10))
        else:
            fig = msg_fig("Unknown plot type")
        fig.tight_layout(); return fig
    except Exception as e:
        return msg_fig(f"⚠️ {e}")
```

## Experiment 4a – Simple & Multiple Linear Regression

```python
REG_TARGETS = ["Rating", "Runtime", "log_Votes", "Year"]
REG_FEATS = NUM_FEATS + ["Rating"] + GENRE_COLS + LANG_COLS

def _reg_metrics(m, Xtr, ytr, Xte, yte):
    ptr, pte = m.predict(Xtr), m.predict(Xte)
    return dict(r2_tr=r2_score(ytr, ptr), r2_te=r2_score(yte, pte), mae=mean_absolute_error(yte, pte),
                rmse=mean_squared_error(yte, pte) ** 0.5, pte=pte)

def _r2_words(r2):
    return "very weak" if r2 < 0.1 else "weak" if r2 < 0.3 else "moderate" if r2 < 0.6 else "strong"

def run_simple_lr(xcol, ycol, test_size, x_new):
    plt.close("all")
    if xcol == ycol: return "⚠️ Pick different columns for X and Y.", None, None
    Xtr, Xte, ytr, yte = train_test_split(df[[xcol]], df[ycol], test_size=test_size, random_state=RS)
    m = LinearRegression().fit(Xtr, ytr); r = _reg_metrics(m, Xtr, ytr, Xte, yte)
    fig, ax = plt.subplots(figsize=(8, 5.5))
    s = Xte.assign(y=yte).sample(min(len(Xte), 3000), random_state=RS)
    ax.scatter(s[xcol], s["y"], alpha=0.35, s=14, label="test data")
    xs = np.linspace(df[xcol].min(), df[xcol].max(), 100)
    ax.plot(xs, m.predict(pd.DataFrame({xcol: xs})), color="red", lw=2.5, label="regression line")
    ax.set_xlabel(xcol); ax.set_ylabel(ycol); ax.set_title(f"Simple Linear Regression: {ycol} ~ {xcol}"); ax.legend()
    fig2, ax2 = plt.subplots(figsize=(8, 4)); res = yte - r["pte"]
    sns.histplot(res, kde=True, ax=ax2, color="#55a868"); ax2.axvline(0, color="k", ls="--"); ax2.set_title("Residuals (actual − predicted) on test set")
    pred = float(m.predict(pd.DataFrame({xcol: [x_new]}))[0])
    md = f"""### Results
**Equation:** `{ycol} = {m.intercept_:.4f} + {m.coef_[0]:.5f} × {xcol}`

| Metric | Value |
|---|---|
| R² (train / test) | {r['r2_tr']:.4f} / {r['r2_te']:.4f} |
| MAE (test) | {r['mae']:.4f} |
| RMSE (test) | {r['rmse']:.4f} |

**🔮 Prediction:** for {xcol} = {x_new} → predicted {ycol} ≈ **{pred:.3f}**

**Conclusion:** `{xcol}` alone gives a **{_r2_words(r['r2_te'])}** fit (R² = {r['r2_te']:.3f}). Each +1 in {xcol} changes {ycol} by about {m.coef_[0]:.4f}."""
    return md, fig, fig2

def run_multiple_lr(target, feats, test_size):
    plt.close("all")
    feats = [f for f in feats if f != target]
    if len(feats) < 2: return "⚠️ Select at least 2 features (other than the target).", None, None, None
    Xtr, Xte, ytr, yte = train_test_split(df[feats], df[target], test_size=test_size, random_state=RS)
    m = LinearRegression().fit(Xtr, ytr); r = _reg_metrics(m, Xtr, ytr, Xte, yte)
    n, p = len(yte), len(feats); adj = 1 - (1 - r["r2_te"]) * (n - 1) / (n - p - 1)
    sc = StandardScaler().fit(Xtr); ms = LinearRegression().fit(sc.transform(Xtr), ytr)
    coef = pd.DataFrame({"feature": feats, "coefficient": m.coef_.round(5), "std_coefficient": ms.coef_.round(4)}) \
             .sort_values("std_coefficient", key=abs, ascending=False)
    fig, axs = plt.subplots(1, 2, figsize=(12, 5))
    axs[0].scatter(yte, r["pte"], alpha=0.3, s=12); lim = [min(yte.min(), r["pte"].min()), max(yte.max(), r["pte"].max())]
    axs[0].plot(lim, lim, "r--"); axs[0].set_xlabel("Actual"); axs[0].set_ylabel("Predicted"); axs[0].set_title("Actual vs Predicted (test)")
    axs[1].scatter(r["pte"], yte - r["pte"], alpha=0.3, s=12); axs[1].axhline(0, color="r", ls="--")
    axs[1].set_xlabel("Predicted"); axs[1].set_ylabel("Residual"); axs[1].set_title("Residual plot"); fig.tight_layout()
    fig2, ax = plt.subplots(figsize=(8, max(3, 0.35 * len(feats) + 1)))
    c = coef.iloc[::-1]; ax.barh(c["feature"], c["std_coefficient"], color=np.where(c["std_coefficient"] >= 0, "#4c72b0", "#c44e52"))
    ax.set_title("Standardised coefficients (relative importance)"); fig2.tight_layout()
    top = coef.iloc[0]
    md = f"""### Results – predicting `{target}` from {len(feats)} features
| Metric | Value |
|---|---|
| R² (train / test) | {r['r2_tr']:.4f} / {r['r2_te']:.4f} |
| Adjusted R² (test) | {adj:.4f} |
| MAE (test) | {r['mae']:.4f} |
| RMSE (test) | {r['rmse']:.4f} |
| Intercept | {m.intercept_:.4f} |

**Conclusion:** the model explains about **{max(r['r2_te'],0)*100:.1f}%** of the variation in {target} ({_r2_words(r['r2_te'])} fit). The most influential feature is **{top['feature']}**."""
    return md, fig, fig2, coef.reset_index(drop=True)
```

## Experiment 4b – Attribute Relevance Analysis + Decision Tree
Attributes are ranked by **information gain** (mutual information with the class). You choose how many top attributes to keep, then the tree is built on those.

```python
_MI = {}
def mi_table(target):
    if target not in _MI:
        f = get_features(target)
        mi = mutual_info_classif(df[f], df[target].astype(str), discrete_features=[c in BIN_COLS or c == "Num_Genres" for c in f], random_state=RS)
        _MI[target] = pd.Series(mi, index=f).sort_values(ascending=False)
    return _MI[target]

def run_tree(target, test_size, criterion, max_depth, min_leaf, top_k):
    plt.close("all")
    mi = mi_table(target); k = int(min(top_k, len(mi))); sel = list(mi.index[:k])
    y = df[target].astype(str)
    Xtr, Xte, ytr, yte = train_test_split(df[sel], y, test_size=test_size, random_state=RS, stratify=y)
    clf = DecisionTreeClassifier(criterion=criterion, max_depth=int(max_depth), min_samples_leaf=int(min_leaf), random_state=RS).fit(Xtr, ytr)
    pred = clf.predict(Xte); acc_te = accuracy_score(yte, pred); acc_tr = accuracy_score(ytr, clf.predict(Xtr))
    base = y.value_counts(normalize=True).max()
    # relevance figure
    fig, axs = plt.subplots(1, 2, figsize=(13, max(4, 0.32 * len(mi) + 1)))
    cols = ["#4c72b0" if f in sel else "#c8c8c8" for f in mi.index[::-1]]
    axs[0].barh(mi.index[::-1], mi.values[::-1], color=cols); axs[0].set_title("Attribute relevance (information gain)\nblue = kept, grey = dropped")
    imp = pd.Series(clf.feature_importances_, index=sel).sort_values()
    axs[1].barh(imp.index, imp.values, color="#55a868"); axs[1].set_title("Importance inside the trained tree"); fig.tight_layout()
    # tree figure
    fig_t, ax = plt.subplots(figsize=(20, 9))
    plot_tree(clf, max_depth=3, feature_names=sel, class_names=list(clf.classes_), filled=True, rounded=True, fontsize=9, proportion=True, ax=ax)
    ax.set_title(f"Decision tree (top 3 levels shown; full depth = {clf.get_depth()}, leaves = {clf.get_n_leaves()})"); fig_t.tight_layout()
    fig_c = cm_fig(confusion_matrix(yte, pred, labels=clf.classes_), clf.classes_, f"Confusion matrix (accuracy {acc_te:.3f})")
    md = f"""### Results – target: `{target}`
| | |
|---|---|
| Attributes kept | {k} of {len(mi)} → `{', '.join(sel)}` |
| Train accuracy | {acc_tr:.4f} |
| **Test accuracy** | **{acc_te:.4f}** |
| Majority-class baseline | {base:.4f} |
| Tree depth / leaves | {clf.get_depth()} / {clf.get_n_leaves()} |

**Conclusion:** the most relevant attribute is **{mi.index[0]}**. The tree reaches {acc_te*100:.1f}% test accuracy vs a {base*100:.1f}% baseline ({'a real improvement' if acc_te > base + 0.02 else 'only marginally better than guessing the majority class'}). {'Train accuracy is much higher than test accuracy → overfitting; lower max depth or raise min leaf size.' if acc_tr - acc_te > 0.08 else 'Train and test accuracy are close → little overfitting.'}"""
    return md, fig, fig_t, fig_c, report_df(yte, pred)
```

## Experiment 4c – Naïve Bayes Classification

```python
def run_nb(target, test_size, smoothing_exp, normalize_cm):
    plt.close("all")
    f = get_features(target); y = df[target].astype(str)
    Xtr, Xte, ytr, yte = train_test_split(df[f], y, test_size=test_size, random_state=RS, stratify=y)
    nb = GaussianNB(var_smoothing=10.0 ** smoothing_exp).fit(Xtr, ytr); pred = nb.predict(Xte)
    acc = accuracy_score(yte, pred); cv = cross_val_score(GaussianNB(var_smoothing=10.0 ** smoothing_exp), df[f], y, cv=5).mean()
    rep = report_df(yte, pred)
    cls = rep[~rep["class"].isin(["accuracy", "macro avg", "weighted avg"])]
    macro = rep[rep["class"] == "macro avg"].iloc[0]
    fig_c = cm_fig(confusion_matrix(yte, pred, labels=nb.classes_), nb.classes_, f"Confusion matrix (accuracy {acc:.3f})", normalize_cm)
    fig, ax = plt.subplots(figsize=(8, 5)); m = cls.melt(id_vars="class", value_vars=["precision", "recall", "f1-score"])
    sns.barplot(data=m, x="class", y="value", hue="variable", ax=ax); ax.set_ylim(0, 1); ax.set_title("Per-class precision / recall / F1")
    plt.setp(ax.get_xticklabels(), rotation=25, ha="right"); fig.tight_layout()
    base = y.value_counts(normalize=True).max()
    md = f"""### Results – Gaussian Naïve Bayes, target: `{target}`
| Metric | Value |
|---|---|
| **Accuracy (test)** | **{acc:.4f}** |
| 5-fold CV accuracy | {cv:.4f} |
| Macro precision | {macro['precision']:.4f} |
| Macro recall | {macro['recall']:.4f} |
| Macro F1-score | {macro['f1-score']:.4f} |
| Majority-class baseline | {base:.4f} |

**Conclusion:** Naïve Bayes gets {acc*100:.1f}% accuracy (baseline {base*100:.1f}%). Weakest class by F1: **{cls.sort_values('f1-score').iloc[0]['class']}** – check the confusion matrix to see which classes it mixes up."""
    return md, fig_c, fig, rep
```

## Experiment 5a – K-Means Clustering

```python
CLUSTER_FEATS = ["Year", "Runtime", "Rating", "log_Votes", "Num_Genres"]
_elbow = {}

def _pca_scatter(Xp, labels, centers, title, var):
    fig, ax = plt.subplots(figsize=(8, 6)); idx = np.random.RandomState(RS).permutation(len(Xp))[:6000]
    sc = ax.scatter(Xp[idx, 0], Xp[idx, 1], c=labels[idx], cmap="tab10", s=14, alpha=0.6)
    if centers is not None: ax.scatter(centers[:, 0], centers[:, 1], c="black", marker="X", s=220, edgecolors="white", label="centroids"); ax.legend()
    ax.set_xlabel(f"PC1 ({var[0]*100:.0f}% var)"); ax.set_ylabel(f"PC2 ({var[1]*100:.0f}% var)"); ax.set_title(title)
    ax.legend(*sc.legend_elements(), title="cluster", loc="best") if centers is None else None
    fig.tight_layout(); return fig

def _profile(data, labels):
    p = data.assign(cluster=labels).groupby("cluster").agg(**{c: (c, "mean") for c in data.columns}).round(2)
    p.insert(0, "size", pd.Series(labels).value_counts().sort_index().values); return p.reset_index()

def _profile_heat(prof, cols):
    z = (prof[cols] - prof[cols].mean()) / prof[cols].std(ddof=0).replace(0, 1)
    fig, ax = plt.subplots(figsize=(7, 1 + 0.6 * len(prof)))
    sns.heatmap(z.set_index(prof["cluster"]), annot=prof[cols].values, fmt=".1f", cmap="coolwarm", center=0, cbar_kws={"label": "relative to other clusters"}, ax=ax)
    ax.set_title("Cluster profile (numbers = mean values, colour = relative level)"); fig.tight_layout(); return fig

def run_kmeans(feats, k, scale, max_k):
    plt.close("all")
    if len(feats) < 2: return "⚠️ Select at least 2 features.", None, None, None, None
    X = df[feats].values; Xs = StandardScaler().fit_transform(X) if scale else X
    key = (tuple(feats), scale, int(max_k))
    if key not in _elbow:
        ks = list(range(2, int(max_k) + 1)); inert, sil = [], []
        for kk in ks:
            m = KMeans(kk, n_init=5, random_state=RS).fit(Xs); inert.append(m.inertia_)
            sil.append(silhouette_score(Xs, m.labels_, sample_size=4000, random_state=RS))
        _elbow[key] = (ks, inert, sil)
    ks, inert, sil = _elbow[key]
    fig, axs = plt.subplots(1, 2, figsize=(11, 4))
    axs[0].plot(ks, inert, "o-"); axs[0].axvline(k, color="r", ls="--"); axs[0].set_title("Elbow method (inertia)"); axs[0].set_xlabel("k")
    axs[1].plot(ks, sil, "s-", color="green"); axs[1].axvline(k, color="r", ls="--"); axs[1].set_title("Silhouette score"); axs[1].set_xlabel("k"); fig.tight_layout()
    km = KMeans(int(k), n_init=10, random_state=RS).fit(Xs); lab = km.labels_
    pca = PCA(2, random_state=RS).fit(Xs); fig_p = _pca_scatter(pca.transform(Xs), lab, pca.transform(km.cluster_centers_), f"K-Means clusters (k={k}) – PCA projection", pca.explained_variance_ratio_)
    prof = _profile(df[feats], lab); fig_h = _profile_heat(prof, feats)
    s = silhouette_score(Xs, lab, sample_size=5000, random_state=RS); best = ks[int(np.argmax(sil))]
    md = f"""### Results – K-Means (k = {k})
| Metric | Value |
|---|---|
| Inertia (WCSS) | {km.inertia_:,.1f} |
| Silhouette score (↑ better) | {s:.4f} |
| Davies-Bouldin (↓ better) | {davies_bouldin_score(Xs, lab):.4f} |
| Calinski-Harabasz (↑ better) | {calinski_harabasz_score(Xs, lab):,.0f} |

**Conclusion:** silhouette is highest at **k = {best}** among the k's tested. The profile heat-map shows what makes each cluster different (e.g. old vs new films, long vs short, popular vs obscure)."""
    return md, fig, fig_p, fig_h, prof
```

## Experiment 5b – Agglomerative (Hierarchical) Clustering
Works on a random sample (hierarchical clustering needs an n × n distance matrix, so it can't use all rows).

```python
def run_agglo(feats, n_clusters, linkage_m, n_sample, scale):
    plt.close("all")
    if len(feats) < 2: return "⚠️ Select at least 2 features.", None, None, None
    s = df.sample(int(n_sample), random_state=RS).reset_index(drop=True); X = s[feats].values
    Xs = StandardScaler().fit_transform(X) if scale else X
    Z = linkage(Xs, method=linkage_m); k = int(n_clusters)
    lab = AgglomerativeClustering(n_clusters=k, linkage=linkage_m).fit_predict(Xs)
    thr = (Z[-(k - 1), 2] + Z[-k, 2]) / 2 if k > 1 else 0
    fig, ax = plt.subplots(figsize=(12, 5))
    dendrogram(Z, truncate_mode="lastp", p=30, color_threshold=thr, show_leaf_counts=True, leaf_rotation=90, ax=ax)
    ax.axhline(thr, color="r", ls="--", label=f"cut for {k} clusters"); ax.legend()
    ax.set_title(f"Dendrogram ({linkage_m} linkage, last 30 merges shown)"); ax.set_ylabel("distance"); fig.tight_layout()
    pca = PCA(2, random_state=RS).fit(Xs); fig_p = _pca_scatter(pca.transform(Xs), lab, None, f"Agglomerative clusters (k={k}, {linkage_m}) – PCA projection", pca.explained_variance_ratio_)
    prof = _profile(s[feats], lab); coph = cophenet(Z, pdist(Xs))[0]
    md = f"""### Results – Agglomerative clustering ({linkage_m} linkage, {len(s)} movies sampled)
| Metric | Value |
|---|---|
| Silhouette score (↑ better) | {silhouette_score(Xs, lab):.4f} |
| Davies-Bouldin (↓ better) | {davies_bouldin_score(Xs, lab):.4f} |
| Cophenetic correlation (how faithful the tree is) | {coph:.4f} |
| Cluster sizes | {', '.join(map(str, np.bincount(lab)))} |

**Conclusion:** the dendrogram shows how clusters merge as distance grows; cutting at the red line gives {k} clusters. {'Single linkage tends to chain points into one giant cluster – try ward/complete.' if linkage_m == 'single' else 'Ward linkage minimises within-cluster variance and usually gives the most balanced clusters.' if linkage_m == 'ward' else ''}"""
    return md, fig, fig_p, prof
```

## Experiment 6 – Apriori (Frequent Itemsets & Association Rules)
Every **movie is a transaction**; its **items** are its genres (and optionally language, rating band, era). Rules such as `{Romance} → {Musical}` show which items occur together.

```python
_TX = {}
def transactions(parts):
    key = tuple(sorted(parts))
    if key not in _TX:
        g = df_all.dropna(subset=["Genre"]).copy(); fr = []
        if "Genres" in parts:   fr.append(g["Genre"].str.get_dummies(sep=", ").add_prefix("Genre: "))
        if "Language" in parts: fr.append(pd.get_dummies(g["Language"], prefix="Lang", prefix_sep=": "))
        if "Rating band" in parts:
            fr.append(pd.get_dummies(pd.cut(g["Rating"], [0, 5.5, 7, 10.1], right=False, labels=["Low", "Medium", "High"]), prefix="Rating", prefix_sep=": "))
        if "Era (decade)" in parts:
            fr.append(pd.get_dummies((g["Year"] // 10 * 10).dropna().astype(int).astype(str) + "s", prefix="Era", prefix_sep=": "))
        _TX[key] = pd.concat(fr, axis=1).fillna(False).astype(bool)
    return _TX[key]

def _fs(s): return ", ".join(sorted(s))

def run_apriori(parts, min_sup, min_conf, min_lift, max_len, top_n):
    plt.close("all")
    if not parts: return "⚠️ Select at least one item type.", None, None, None, None
    T = transactions(parts)
    fi = apriori(T, min_support=min_sup, use_colnames=True, max_len=int(max_len))
    if fi.empty: return f"⚠️ No itemsets with support ≥ {min_sup}. Lower the minimum support.", None, None, None, None
    fi["size"] = fi["itemsets"].apply(len)
    rules = pd.DataFrame()
    if (fi["size"] >= 2).any():
        rules = association_rules(fi, metric="confidence", min_threshold=min_conf)
        rules = rules[rules["lift"] >= min_lift].sort_values("lift", ascending=False)
    top_n = int(top_n)
    fi_show = fi.sort_values("support", ascending=False).head(top_n).copy()
    fi_show["itemsets"] = fi_show["itemsets"].apply(_fs); fi_show["support"] = fi_show["support"].round(4)
    fig, ax = plt.subplots(figsize=(9, max(4, 0.3 * len(fi_show) + 1)))
    ax.barh(fi_show["itemsets"][::-1], fi_show["support"][::-1], color="#4c72b0"); ax.set_xlabel("support"); ax.set_title(f"Top {len(fi_show)} frequent itemsets"); fig.tight_layout()
    if rules.empty:
        return (f"### Results\n{len(T):,} transactions · **{len(fi)}** frequent itemsets · **0 rules** pass the filters – lower min confidence / lift or raise max length.",
                fig, fi_show[["itemsets", "support"]], pd.DataFrame(), msg_fig("No rules to plot"))
    r = rules.copy()
    fig2, ax2 = plt.subplots(figsize=(8, 5.5))
    sc = ax2.scatter(r["support"], r["confidence"], c=r["lift"], cmap="viridis", s=45, alpha=0.8)
    fig2.colorbar(sc, label="lift"); ax2.set_xlabel("support"); ax2.set_ylabel("confidence"); ax2.set_title(f"{len(r)} association rules"); fig2.tight_layout()
    show = r.head(top_n).copy(); show["antecedents"] = show["antecedents"].apply(_fs); show["consequents"] = show["consequents"].apply(_fs)
    show = show[["antecedents", "consequents", "support", "confidence", "lift", "leverage"]].round(4)
    b = show.iloc[0]
    md = f"""### Results
**{len(T):,}** transactions · **{len(fi)}** frequent itemsets (support ≥ {min_sup}) · **{len(r)}** rules (confidence ≥ {min_conf}, lift ≥ {min_lift})

**Conclusion:** the strongest rule by lift is **{{{b['antecedents']}}} → {{{b['consequents']}}}** (support {b['support']:.3f}, confidence {b['confidence']:.2f}, lift {b['lift']:.2f}). Lift > 1 means the items appear together more often than by chance; lift ≈ 1 means no real association."""
    return md, fig, fi_show[["itemsets", "support"]], show, fig2
```

## 🎯 Movie Recommender
Pick your **genres**, **minimum rating**, and optional filters (language, year, popularity) and get a ranked list of movies.

**How it works (content-based filtering):**
1. **Hard filters** remove movies that don't meet your rating / votes / language / year limits.
2. **Genre match:** *coverage* = fraction of your chosen genres that the movie has (so a Drama+Romance movie scores 100% for a Drama+Romance request).
3. **Weighted rating** (IMDb-style) pulls ratings of movies with few votes toward the overall average, so a 10/10 film with 5 votes can't beat a well-reviewed film with thousands:
   `WR = v/(v+m) × R + m/(v+m) × C` (R = movie rating, v = votes, m = 200, C = mean rating of all movies).
4. **Match score** = `0.6 × genre coverage + 0.4 × (WR / 10)`.

```python
REC = df_all.dropna(subset=["Rating", "Genre"]).copy()          # every movie that has a rating and a genre
REC["Votes"] = REC["Votes"].fillna(0)
REC["Key"] = REC["ID"].fillna(REC["Movie Name"] + "|" + REC["Year"].astype(str))   # same film listed under several languages = one film
REC["GenreSet"] = REC["Genre"].str.split(r",\s*").apply(frozenset)
_C, _M = REC["Rating"].mean(), 200
REC["WRating"] = REC["Votes"] / (REC["Votes"] + _M) * REC["Rating"] + _M / (REC["Votes"] + _M) * _C

ALL_GENRES   = [g for g, c in REC["Genre"].str.split(r",\s*").explode().value_counts().items() if c >= 30]
LANG_CHOICES = [(l.title(), l) for l in REC["Language"].value_counts().index]
YEAR_MIN, YEAR_MAX = int(REC["Year"].min()), int(REC["Year"].max())
REC_RANKS = ["Best match", "Highest rating", "Most popular (votes)"]

def run_recommend(genres, mode, min_rating, langs, y_from, y_to, min_votes, rank_by, n):
    plt.close("all")
    n = int(n); sel = set(genres or []); r = REC
    funnel = [("all rated movies", len(r))]
    r = r[r["Rating"] >= min_rating];            funnel.append((f"rating ≥ {min_rating:g}", len(r)))
    r = r[r["Votes"] >= min_votes];              funnel.append((f"votes ≥ {int(min_votes):,}", len(r)))
    if langs:
        r = r[r["Language"].isin(langs)];        funnel.append(("language", len(r)))
    if y_from > YEAR_MIN or y_to < YEAR_MAX:
        r = r[r["Year"].between(y_from, y_to)];  funnel.append((f"year {int(y_from)}–{int(y_to)}", len(r)))
    r = r.copy()
    if sel and len(r):
        r["Matched"] = r["GenreSet"].apply(lambda s: s & sel)
        r["Coverage"] = r["Matched"].apply(len) / len(sel)
        r = r[r["Coverage"] == 1] if mode == "All selected genres" else r[r["Coverage"] > 0]
        funnel.append(("genre match", len(r)))
    else:
        r["Matched"] = [frozenset()] * len(r); r["Coverage"] = 1.0
    trail = " → ".join(f"{lab}: **{cnt:,}**" for lab, cnt in funnel)
    if r.empty:
        return (f"### 😕 No movies match\n{trail}\n\nTry lowering the minimum rating or votes, widening the years, "
                f"or switching genre match to *Any selected genre*."), pd.DataFrame(), None
    r["Score"] = 0.6 * r["Coverage"] + 0.4 * r["WRating"] / 10
    keys = {"Best match": ["Score", "Votes"], "Highest rating": ["Rating", "Votes"], "Most popular (votes)": ["Votes", "Rating"]}[rank_by]
    r = r.sort_values(keys, ascending=False)
    langs_by = r.groupby("Key")["Language"].agg(lambda x: ", ".join(sorted({v.title() for v in x})))
    r = r.drop_duplicates("Key").head(n).reset_index(drop=True)
    r["LangText"] = r["Key"].map(langs_by)
    yr = lambda v: "—" if pd.isna(v) else str(int(v))
    out = pd.DataFrame({"Rank": np.arange(1, len(r) + 1), "Movie": r["Movie Name"], "Year": r["Year"].apply(yr),
                        "Genre": r["Genre"], "Language": r["LangText"], "Rating": r["Rating"],
                        "Votes": r["Votes"].astype(int), "Match %": (r["Score"] * 100).round().astype(int)})
    cards = []
    for i, row in r.head(5).iterrows():
        why = f" · ✅ matches: {', '.join(sorted(row['Matched']))}" if sel else ""
        cards.append(f"**{i + 1}. {row['Movie Name']}** ({yr(row['Year'])}) — ⭐ {row['Rating']:.1f} · {int(row['Votes']):,} votes  \n"
                     f"🎭 {row['Genre']} · 🗣️ {row['LangText']}{why}")
    md = f"### 🎯 Top {len(r)} recommendations (ranked by: {rank_by})\n" + "\n\n".join(cards) + f"\n\n*Filter funnel – {trail}*"
    fig, ax = plt.subplots(figsize=(9, max(3.5, 0.38 * len(out) + 1)))
    labels = [f"{a[:34]} ({b})" for a, b in zip(out["Movie"], out["Year"])]
    bars = ax.barh(labels[::-1], out["Rating"][::-1], color=plt.cm.viridis(out["Match %"][::-1] / 100))
    for b_, v in zip(bars, out["Votes"][::-1]): ax.text(b_.get_width() + 0.05, b_.get_y() + b_.get_height() / 2, f"{v:,} votes", va="center", fontsize=8)
    ax.set_xlim(0, 11); ax.set_xlabel("IMDb rating (bar colour = match score, darker→brighter = better)"); ax.set_title("Recommended movies"); fig.tight_layout()
    return md, out, fig
```

```python
HEADER = f"""# 🎬 Indian Movies – Data Mining Lab
**Dataset:** {SRC} · **{len(df_all):,}** movies · **{len(df):,}** complete rows used for modelling.
Use the tabs to switch experiments. Change the controls on the left and press **Run**. The last tab is the 🎯 **Movie Recommender**."""

def viva(text):
    with gr.Accordion("🎓 Viva notes (plain language)", open=False): gr.Markdown(text)

def sl(label, lo, hi, val, step=1): return gr.Slider(lo, hi, value=val, step=step, label=label)

with gr.Blocks(title="Indian Movies – Data Mining Lab") as demo:
    gr.Markdown(HEADER)
    with gr.Tabs():
        # ---------------- Dataset ----------------
        with gr.Tab("📁 Dataset"):
            view = gr.Radio(["Raw file (first rows)", "Cleaned data", "Summary statistics", "Missing values"], value="Cleaned data", label="View")
            nrows = sl("Rows to show", 5, 100, 20)
            tbl = gr.Dataframe(interactive=False, wrap=True)
            def show_data(v, n):
                n = int(n)
                if v.startswith("Raw"): return raw.head(n)
                if v == "Cleaned data": return df[["ID", "Movie Name", "Year", "Runtime", "Rating", "Votes", "Genre", "Language", "Rating_Category"]].head(n)
                if v == "Summary statistics": return df[NUM_COLS].describe().round(2).reset_index().rename(columns={"index": "stat"})
                return pd.DataFrame({"column": MISSING_PCT.index, "% missing (original file)": MISSING_PCT.values})
            view.change(show_data, [view, nrows], tbl); nrows.change(show_data, [view, nrows], tbl); demo.load(show_data, [view, nrows], tbl)
            gr.Markdown("**Cleaning steps:** `-` → missing · Year/Runtime/Votes parsed to numbers · duplicates removed · runtimes outside 40–300 min dropped · `log_Votes`, `Num_Genres`, genre & language flags added · `Rating_Category` = Low (<5.5) / Medium (5.5–7) / High (≥7).")

        # ---------------- Exp 3 ----------------
        with gr.Tab("3 · Visualization"):
            with gr.Row():
                with gr.Column(scale=1):
                    v_kind = gr.Dropdown(VIZ_TYPES, value="Histogram", label="Plot type")
                    v_x = gr.Dropdown(NUM_COLS + [c for c in CAT_COLS if c != "Num_Genres"], value="Rating", label="X column")
                    v_y = gr.Dropdown(["(none)"] + NUM_COLS, value="(none)", label="Y column (numeric, optional)")
                    v_h = gr.Dropdown(HUE_CHOICES, value="(none)", label="Colour by")
                    v_bins = sl("Histogram bins", 5, 80, 30); v_top = sl("Top-N categories", 3, 20, 10)
                    v_btn = gr.Button("Draw plot", variant="primary")
                with gr.Column(scale=3):
                    v_out = gr.Plot(label="Plot")
            gr.Markdown("*Tips: Histogram → X=Rating · Box/Violin → X=Language, Y=Rating · Scatter → X=log_Votes, Y=Rating · Line → X=Year, Y=Rating (or none for counts) · Bar → X=Primary_Genre, Y=Rating.*")
            viva("Visualization helps us *see* the data before modelling. **Histogram** = shape of one variable; **box/violin** = spread, median and outliers per group; **scatter/hexbin** = relationship between two numbers; **line** = trend over time; **bar/pie** = category counts; **heatmap** = which numeric columns move together (correlation); **pair plot** = all pairs at once; **missing-value chart** = data quality.")
            ins = [v_kind, v_x, v_y, v_h, v_bins, v_top]
            v_btn.click(run_viz, ins, v_out); demo.load(run_viz, ins, v_out)

        # ---------------- Exp 4a ----------------
        with gr.Tab("4a · Linear Regression"):
            with gr.Tabs():
                with gr.Tab("Simple LR"):
                    with gr.Row():
                        with gr.Column(scale=1):
                            s_x = gr.Dropdown(NUM_FEATS, value="log_Votes", label="Predictor X"); s_y = gr.Dropdown(REG_TARGETS, value="Rating", label="Target Y")
                            s_t = sl("Test size", 0.1, 0.5, 0.25, 0.05); s_n = gr.Number(value=3.0, label="Try a value of X to predict Y")
                            s_btn = gr.Button("Run Simple LR", variant="primary")
                        with gr.Column(scale=3):
                            s_md = gr.Markdown(); s_p1 = gr.Plot(); s_p2 = gr.Plot()
                    s_in = [s_x, s_y, s_t, s_n]; s_out = [s_md, s_p1, s_p2]; OUT_COUNT["run_simple_lr"] = 3
                    s_btn.click(safe(run_simple_lr), s_in, s_out); demo.load(safe(run_simple_lr), s_in, s_out)
                with gr.Tab("Multiple LR"):
                    with gr.Row():
                        with gr.Column(scale=1):
                            m_y = gr.Dropdown(REG_TARGETS, value="Rating", label="Target Y")
                            m_f = gr.CheckboxGroup(REG_FEATS, value=["Year", "Runtime", "log_Votes", "Num_Genres"] + GENRE_COLS[:5], label="Features (target is ignored if ticked)")
                            m_t = sl("Test size", 0.1, 0.5, 0.25, 0.05); m_btn = gr.Button("Run Multiple LR", variant="primary")
                        with gr.Column(scale=3):
                            m_md = gr.Markdown(); m_p1 = gr.Plot(); m_p2 = gr.Plot(); m_tb = gr.Dataframe(label="Coefficients", interactive=False)
                    m_in = [m_y, m_f, m_t]; m_out = [m_md, m_p1, m_p2, m_tb]; OUT_COUNT["run_multiple_lr"] = 4
                    m_btn.click(safe(run_multiple_lr), m_in, m_out); demo.load(safe(run_multiple_lr), m_in, m_out)
            viva("**Linear regression** fits a straight line (or plane) to predict a *number*. **Simple** = one input, **Multiple** = several inputs. **R²** = fraction of variation explained (0 = useless, 1 = perfect). **MAE/RMSE** = average prediction error (RMSE punishes big errors more). **Residuals** should look random around 0; patterns mean the straight-line assumption is weak. **Standardised coefficients** let you compare importance across features with different units.")

        # ---------------- Exp 4b ----------------
        with gr.Tab("4b · Decision Tree"):
            with gr.Row():
                with gr.Column(scale=1):
                    t_y = gr.Dropdown(TARGET_CHOICES, value="Rating_Category", label="Class to predict")
                    t_c = gr.Radio(["gini", "entropy"], value="gini", label="Split criterion")
                    t_d = sl("Max depth", 1, 15, 5); t_l = sl("Min samples per leaf", 1, 200, 20)
                    t_k = sl("Keep top-K relevant attributes", 2, 21, 12); t_t = sl("Test size", 0.1, 0.5, 0.25, 0.05)
                    t_btn = gr.Button("Run Decision Tree", variant="primary")
                with gr.Column(scale=3):
                    t_md = gr.Markdown(); t_p1 = gr.Plot(label="Attribute relevance"); t_p2 = gr.Plot(label="Tree"); t_p3 = gr.Plot(label="Confusion matrix"); t_tb = gr.Dataframe(label="Classification report", interactive=False)
            t_in = [t_y, t_t, t_c, t_d, t_l, t_k]; t_out = [t_md, t_p1, t_p2, t_p3, t_tb]; OUT_COUNT["run_tree"] = 5
            t_btn.click(safe(run_tree), t_in, t_out); demo.load(safe(run_tree), t_in, t_out)
            viva("A **decision tree** asks a chain of yes/no questions (e.g. 'votes > 100?') to reach a class. **Attribute relevance / information gain** measures how much knowing an attribute reduces uncertainty about the class – we keep the best ones. **Gini/entropy** are two ways to measure how 'mixed' a node is; the tree splits to make nodes purer. **Overfitting** = memorising training data (train accuracy ≫ test accuracy) – controlled by max depth and min leaf size.")

        # ---------------- Exp 4c ----------------
        with gr.Tab("4c · Naïve Bayes"):
            with gr.Row():
                with gr.Column(scale=1):
                    n_y = gr.Dropdown(TARGET_CHOICES, value="Rating_Category", label="Class to predict")
                    n_t = sl("Test size", 0.1, 0.5, 0.25, 0.05); n_s = sl("Variance smoothing (10^x)", -9, -1, -9)
                    n_n = gr.Checkbox(False, label="Normalise confusion matrix (row %)"); n_btn = gr.Button("Run Naïve Bayes", variant="primary")
                with gr.Column(scale=3):
                    n_md = gr.Markdown(); n_p1 = gr.Plot(label="Confusion matrix"); n_p2 = gr.Plot(label="Precision / Recall / F1"); n_tb = gr.Dataframe(label="Classification report", interactive=False)
            n_in = [n_y, n_t, n_s, n_n]; n_out = [n_md, n_p1, n_p2, n_tb]; OUT_COUNT["run_nb"] = 4
            n_btn.click(safe(run_nb), n_in, n_out); demo.load(safe(run_nb), n_in, n_out)
            viva("**Naïve Bayes** uses Bayes' theorem: pick the class with the highest probability given the features, *assuming features are independent* (that's the 'naïve' part). **Accuracy** = % correct. **Precision** = of those predicted as class A, how many really are A. **Recall** = of all real A's, how many we found. **F1** = balance of precision and recall. The **confusion matrix** shows exactly which classes get confused.")

        # ---------------- Exp 5a ----------------
        with gr.Tab("5a · K-Means"):
            with gr.Row():
                with gr.Column(scale=1):
                    k_f = gr.CheckboxGroup(CLUSTER_FEATS, value=CLUSTER_FEATS, label="Features")
                    k_k = sl("Number of clusters (k)", 2, 10, 4); k_m = sl("Max k for elbow plot", 4, 12, 9)
                    k_s = gr.Checkbox(True, label="Standardise features"); k_btn = gr.Button("Run K-Means", variant="primary")
                with gr.Column(scale=3):
                    k_md = gr.Markdown(); k_p1 = gr.Plot(label="Elbow / silhouette"); k_p2 = gr.Plot(label="Clusters"); k_p3 = gr.Plot(label="Profile"); k_tb = gr.Dataframe(label="Cluster profile", interactive=False)
            k_in = [k_f, k_k, k_s, k_m]; k_out = [k_md, k_p1, k_p2, k_p3, k_tb]; OUT_COUNT["run_kmeans"] = 5
            k_btn.click(safe(run_kmeans), k_in, k_out); demo.load(safe(run_kmeans), k_in, k_out)
            viva("**K-Means** puts each point in the nearest of *k* centres, moves each centre to the mean of its points, and repeats until stable. We must choose k: the **elbow method** looks for where adding clusters stops reducing the error (inertia) much; **silhouette** (−1 to 1) says how well points fit their own cluster vs the next one. **Standardising** stops big-scale columns (like Year) dominating. **PCA** squashes many features into 2 axes just so we can draw the clusters.")

        # ---------------- Exp 5b ----------------
        with gr.Tab("5b · Agglomerative"):
            with gr.Row():
                with gr.Column(scale=1):
                    a_f = gr.CheckboxGroup(CLUSTER_FEATS, value=CLUSTER_FEATS, label="Features")
                    a_k = sl("Number of clusters", 2, 10, 4); a_l = gr.Dropdown(["ward", "complete", "average", "single"], value="ward", label="Linkage")
                    a_n = sl("Sample size", 200, 3000, 1000, 100); a_s = gr.Checkbox(True, label="Standardise features"); a_btn = gr.Button("Run Agglomerative", variant="primary")
                with gr.Column(scale=3):
                    a_md = gr.Markdown(); a_p1 = gr.Plot(label="Dendrogram"); a_p2 = gr.Plot(label="Clusters"); a_tb = gr.Dataframe(label="Cluster profile", interactive=False)
            a_in = [a_f, a_k, a_l, a_n, a_s]; a_out = [a_md, a_p1, a_p2, a_tb]; OUT_COUNT["run_agglo"] = 4
            a_btn.click(safe(run_agglo), a_in, a_out); demo.load(safe(run_agglo), a_in, a_out)
            viva("**Agglomerative clustering** is bottom-up: start with every point alone, repeatedly merge the two closest clusters until one remains. The **dendrogram** is the merge history; cutting it at a height gives the clusters. **Linkage** defines 'closest': *ward* (minimise variance), *complete* (farthest points), *average*, *single* (nearest points – can chain). Unlike K-Means you don't need to pick k in advance, but it is slow on big data, so we sample.")

        # ---------------- Exp 6 ----------------
        with gr.Tab("6 · Apriori"):
            with gr.Row():
                with gr.Column(scale=1):
                    p_i = gr.CheckboxGroup(["Genres", "Language", "Rating band", "Era (decade)"], value=["Genres", "Language", "Rating band"], label="Items in each movie-transaction")
                    p_s = sl("Min support", 0.005, 0.3, 0.03, 0.005); p_c = sl("Min confidence", 0.1, 1.0, 0.5, 0.05); p_l = sl("Min lift", 1.0, 5.0, 1.0, 0.1)
                    p_m = sl("Max itemset length", 2, 5, 3); p_n = sl("Rows / bars to show", 5, 50, 15); p_btn = gr.Button("Run Apriori", variant="primary")
                with gr.Column(scale=3):
                    p_md = gr.Markdown(); p_p1 = gr.Plot(label="Frequent itemsets"); p_t1 = gr.Dataframe(label="Frequent itemsets", interactive=False)
                    p_t2 = gr.Dataframe(label="Association rules (sorted by lift)", interactive=False); p_p2 = gr.Plot(label="Rules: support vs confidence")
            p_in = [p_i, p_s, p_c, p_l, p_m, p_n]; p_out = [p_md, p_p1, p_t1, p_t2, p_p2]; OUT_COUNT["run_apriori"] = 5
            p_btn.click(safe(run_apriori), p_in, p_out); demo.load(safe(run_apriori), p_in, p_out)
            viva("**Apriori** finds items that frequently occur together. **Support** = how often an itemset appears (fraction of transactions). **Confidence** of A→B = of the movies containing A, the fraction that also contain B. **Lift** = confidence ÷ how common B is anyway; lift > 1 means a real positive association. Apriori's trick: if an itemset is infrequent, every bigger set containing it is also infrequent, so it prunes them early.")

        # ---------------- Recommender ----------------
        with gr.Tab("🎯 Recommender"):
            with gr.Row():
                with gr.Column(scale=1):
                    r_g = gr.Dropdown(ALL_GENRES, value=["Drama", "Romance"], multiselect=True, label="Favourite genres (pick any number)")
                    r_m = gr.Radio(["Any selected genre", "All selected genres"], value="Any selected genre", label="Genre match")
                    r_r = sl("Minimum rating (out of 10)", 1, 10, 7, 0.5)
                    r_v = sl("Minimum votes (ignore rarely-rated movies)", 0, 5000, 100, 10)
                    r_l = gr.Dropdown(LANG_CHOICES, value=[], multiselect=True, label="Languages (empty = all)")
                    r_y1 = sl("From year", YEAR_MIN, YEAR_MAX, YEAR_MIN); r_y2 = sl("To year", YEAR_MIN, YEAR_MAX, YEAR_MAX)
                    r_k = gr.Radio(REC_RANKS, value="Best match", label="Rank by")
                    r_n = sl("How many recommendations", 3, 30, 10)
                    r_btn = gr.Button("Recommend movies 🎬", variant="primary")
                with gr.Column(scale=3):
                    r_md = gr.Markdown(); r_p = gr.Plot(label="Recommended movies"); r_tb = gr.Dataframe(label="Full list", interactive=False, wrap=True)
            r_in = [r_g, r_m, r_r, r_l, r_y1, r_y2, r_v, r_k, r_n]; r_out = [r_md, r_tb, r_p]; OUT_COUNT["run_recommend"] = 3
            r_btn.click(safe(run_recommend), r_in, r_out); demo.load(safe(run_recommend), r_in, r_out)
            viva("This is **content-based filtering**: we recommend items whose *attributes* (genre, rating, language, year) match what you asked for. **Filters** first remove movies that fail your limits; then each remaining movie gets a **match score** = 60% genre coverage + 40% weighted rating. The **weighted rating** (IMDb formula) shrinks ratings of movies with few votes toward the overall average, so rare 10/10 films don't flood the list. A *collaborative* recommender would instead use many users' ratings – this dataset has no per-user data, so content-based is the right choice.")

demo.queue().launch(share=True, debug=True)   # if the public link fails in your network, use: demo.launch(debug=True)
```
